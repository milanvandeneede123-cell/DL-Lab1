```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_learning_curves(csv_path="01_overfit_model_history.csv"):
    # 1. Lees de data in met Pandas
    try:
        history_df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"Fout: Kan het bestand {csv_path} niet vinden. Zorg dat het in dezelfde map staat.")
        return

    # De CSV-logger van Keras slaat de epochs standaard op als 0, 1, 2...
    # We tellen er 1 bij op zodat de grafiek netjes bij Epoch 1 begint.
    epochs = history_df['epoch'] + 1

    # 2. Maak een figuur aan met 2 subplots (Loss links, Accuracy rechts)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

    # --- Plot 1: Cross-entropy Loss ---
    ax1.plot(epochs, history_df['loss'], label='Training Loss', color='green', linewidth=2)
    ax1.plot(epochs, history_df['val_loss'], label='Validation Loss', color='red', linewidth=2)
    ax1.set_title('Learning Curves: Loss (Regularized)')
    ax1.set_xlabel('Epochs')
    ax1.set_ylabel('Cross-entropy Loss')
    ax1.legend()
    ax1.grid(True, linestyle='--', alpha=0.7)

    # --- Plot 2: Accuracy ---
    ax2.plot(epochs, history_df['accuracy'], label='Training Accuracy', color='green', linewidth=2)
    ax2.plot(epochs, history_df['val_accuracy'], label='Validation Accuracy', color='red', linewidth=2)
    ax2.set_title('Learning Curves: Accuracy (Regularized)')
    ax2.set_xlabel('Epochs')
    ax2.set_ylabel('Accuracy')
    ax2.legend()
    ax2.grid(True, linestyle='--', alpha=0.7)

    # 3. Layout optimaliseren en plot opslaan/tonen
    plt.tight_layout()
    
    # Sla de afbeelding op voor in je verslag!
    plt.savefig('01_underfit_plot.png', dpi=300)
    print("Plot succesvol opgeslagen als '01_underfit_plot.png'")
    
    # Toon de plot op je scherm
    plt.show()

if __name__ == "__main__":
    plot_learning_curves()
```

```python
plot_learning_curves("01_regularized_model_history.csv")
```
